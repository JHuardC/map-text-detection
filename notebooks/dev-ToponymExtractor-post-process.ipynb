# Check post-processing of ToponymExtractor predictions

```python
# Imports
from typing import Final
from pathlib import Path
from dotenv import find_dotenv, load_dotenv
from os import getenv
import geopandas as gp
from outputs import ProcessToponymExtractorPredictions

# Paths
PROJECT_DIR: Final[Path] = Path(find_dotenv(".env", 1, 1)).absolute().parent
load_dotenv(PROJECT_DIR.joinpath(".env"))
LOCAL_DIR: Final[Path] = Path(getenv("LOCAL_DIR"))
```

## Get example predictions

```python
# Load predictions
pred_idx = 11
predictions =\
    [*LOCAL_DIR.glob("outputs/toponym-extractor-polygon-masks/*.gpkg")]
pred_ex = gp.read_file(predictions[pred_idx]) # example file
pred_ex["geometry"] = pred_ex.geometry.buffer(0)
pred_ex.head()
```

## Get control points geodataframe

```python
control_points =\
    gp.read_file(LOCAL_DIR.joinpath("outputs/pngs/control-points.gpkg"))
control_points.head()
```

## Post-process toponym outputs

```python
post_processor = ProcessToponymExtractorPredictions(
    ctrl_points = control_points,
    tiff_dir = LOCAL_DIR.joinpath("data/tiffs"),
    img_h = 2000,
    img_w = 2000
)
outputs = post_processor.process_predictions(
    tiff_fn = f"{predictions[pred_idx].stem}.tif", predictions = pred_ex
)
```
